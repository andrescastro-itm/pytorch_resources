# Introducción a Tensores en PyTorch

## ¿Qué son los Tensores?

Los tensores son la estructura de datos fundamental en PyTorch, similar a los arrays de NumPy, pero con capacidades adicionales para computación en GPU y diferenciación automática.

```python
# Importamos las bibliotecas necesarias
import torch
import numpy as np

# Creación de tensores desde diferentes fuentes

# 1. Desde una lista
lista = [1, 2, 3, 4]
tensor_desde_lista = torch.tensor(lista)
print("Tensor desde lista:", tensor_desde_lista)

# 2. Desde un array de NumPy
array_numpy = np.array([1, 2, 3, 4])
tensor_desde_numpy = torch.from_numpy(array_numpy)
print("Tensor desde NumPy:", tensor_desde_numpy)
```

## Funciones de Creación de Tensores en PyTorch

PyTorch ofrece múltiples funciones para crear tensores con diferentes inicializaciones:

```python
# Funciones de creación de tensores

# Tensor de unos
unos = torch.ones(3, 3)
print("Tensor de unos:\n", unos)

# Tensor de ceros
ceros = torch.zeros(2, 4)
print("\nTensor de ceros:\n", ceros)

# Tensor con valores aleatorios
aleatorio = torch.rand(3, 3)  # Distribución uniforme entre 0 y 1
print("\nTensor aleatorio:\n", aleatorio)

# Tensor con valores aleatorios de distribución normal
normal = torch.randn(3, 3)  # Media 0, varianza 1
print("\nTensor de distribución normal:\n", normal)
```

## Accediendo a Elementos de un Tensor

Exploraremos diferentes formas de acceder y manipular elementos de tensores:

```python
# Creamos un tensor de ejemplo
tensor = torch.tensor([[1, 2, 3], [4, 5, 6]])

# Acceso a elementos individuales
print("Elemento en (0,1):", tensor[0, 1])  # Acceso por índices

# Uso de .item()
# .item() solo funciona con tensores de un solo elemento
tensor_un_elemento = torch.tensor([42])
valor = tensor_un_elemento.item()  # Convierte a valor escalar
print("\nValor con .item():", valor)

# NOTA: .item() NO funciona con tensores de múltiples elementos
# Esto generaría un error
# tensor[0].item()  # Incorrecto para tensores multi-elemento
```

## Atributo requires_grad y Diferenciación Automática

El atributo `requires_grad` es crucial para el aprendizaje profundo, permitiendo el cálculo automático de gradientes:

```python
# Tensores con seguimiento de gradiente

# Tensor sin seguimiento de gradiente
x = torch.tensor([1.0, 2.0, 3.0])
print("requires_grad por defecto:", x.requires_grad)

# Tensor con seguimiento de gradiente
y = torch.tensor([1.0, 2.0, 3.0], requires_grad=True)
print("requires_grad activado:", y.requires_grad)

# Ejemplo de cálculo de gradiente
z = y * 2
loss = z.sum()
loss.backward()  # Calcula gradientes

print("\nGradientes de y:", y.grad)
```

## Conclusiones

- Los tensores son la base de los cálculos en PyTorch
- Pueden crearse desde listas, NumPy arrays o funciones específicas
- `requires_grad` permite el seguimiento de operaciones para aprendizaje automático
- `.item()` solo funciona para tensores de un solo elemento
